```python
def twosum_brute(nums:list[int], target : int):
  for i in range (len(nums)):
    for j in range (i+1, len(nums)):
      if nums[i] + nums[j] == target:
        return [i, j]
  return []
twosum_brute([15,1,7,28,4,20,24,5,26,2,3,14,10,11,18],52)
```

```python
import time
start = time.perf_counter()
twosum_brute([15,1,7,28,4,20,24,5,26,2,3,14,10,11,18],52)
end = time.perf_counter()
print(f"brute force method{end-start:.6f}s")
```

```python
def twosum_map(nums:list[int], target : int):
  map = {}
  for i in range (len(nums)):
    comp = target - nums[i]
    if comp in map:
      return [map[comp], i]
    map[i]  = comp
```

```python
import time
start1 = time.perf_counter()
twosum_map([15,1,7,28,4,20,24,5,26,2,3,14,10,11,18],52)
end1 = time.perf_counter()
print(f"map force method{end-start:.6f}s")
```
